```python
# ==========================================
# Part 1: NumPy Array Operations
# ==========================================

# 1. Create a 1D NumPy Array
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("1D Array:", temperatures_w1)

# 2. Inspection and Properties
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)

# 3. Array Operations
temperatures_fahrenheit = (temperatures_w1 * 9/5) + 32
print("Fahrenheit:", temperatures_fahrenheit)
print("Maximum:", np.max(temperatures_w1))
print("Minimum:", np.min(temperatures_w1))
print("Mean:", np.mean(temperatures_w1))

# 4. Array Slicing and Indexing
print("First 3 days:", temperatures_w1[:3])
print("Weekend:", temperatures_w1[-2:])
print("Middle 3 days:", temperatures_w1[2:5])

# 5. Create a 2D Array
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]   # Week 2
])
print("2D Array:\n", temperatures)

# 6. Inspect and Slice the 2D Array
print("2D Shape:", temperatures.shape)
print("2D Data Type:", temperatures.dtype)
print("Total Elements:", temperatures.size)
print("Week 1:", temperatures[0])
print("Week 2:", temperatures[1])
print("Week 1 Weekend:", temperatures[0, -2:])
print("Week 2 Weekend:", temperatures[1, -2:])
```

```python
# ==========================================
# Part 2: Pandas Series
# ==========================================

# 1. Creating Pandas Series
marks = pd.Series([95, 92, 89, 85, 80], index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)

# 2. Indexing and Slicing
print("1st Rank Mark:", marks.iloc[0])
print("Top 3 Ranks:\n", marks.loc['Rank1':'Rank3'])
print("3rd Rank Mark:", marks.loc['Rank3'])
print("Marks > 90:\n", marks[marks > 90])

# 3. Manipulating Series
marks.loc['Rank1'] = 100
marks = marks.drop('Rank5')
cgpa = marks / 10
print("CGPA:\n", cgpa)
```

```python
# ==========================================
# Part 3: Pandas DataFrame
# ==========================================

# 1. Creating Pandas DataFrame
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'East', 'West', 'South', 'North', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}
transactions = pd.DataFrame(data)
print(transactions)

# 2. Data Exploration
print("Head:\n", transactions.head())
print("Tail:\n", transactions.tail())
print("Shape:", transactions.shape)
print("Columns:", transactions.columns)
print("Data Types:\n", transactions.dtypes)
print(transactions[['ProductCategory', 'Amount']])
print(transactions.iloc[:, -3:])
print(transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)])
print("Value Counts:\n", transactions['ProductCategory'].value_counts())
print("Unique Regions:", transactions['Region'].unique())
print("Mean Amount by Region:\n", transactions.groupby('Region')['Amount'].mean())

# 3. Manipulating the DataFrame
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
transactions['Discount'] = transactions['Amount'] * 0.10
transactions = transactions[transactions['TransactionID'] != 109]
transactions = transactions.drop(columns=['Discount'])
print(transactions)
```
